# Validation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
cointegration_pairs = pd.read_csv('../data/manipulated/cointegrated_pairs.csv')
SPREADS_DIR  = '../data/manipulated/spreads/'
VAL_START    = '2024-01-02'
VAL_END      = '2025-01-01'

# use train + val returns index to fix integer-indexed spread CSVs
train_returns = pd.read_csv('../data/raw/equity_returns_train.csv', index_col=0, parse_dates=True)
val_returns   = pd.read_csv('../data/raw/equity_returns_val.csv',   index_col=0, parse_dates=True)
test_returns  = pd.read_csv('../data/raw/equity_returns_test.csv',  index_col=0, parse_dates=True)
date_index    = train_returns.index.append(val_returns.index).append(test_returns.index)

print(f'date_index length: {len(date_index)}')
print(f'date_index range:  {date_index[0].date()} to {date_index[-1].date()}')
print(f'Validation period: {VAL_START} to {VAL_END}')
```

```python
def load_spread(filename):
    """
    Load a spread CSV and restore the date index from residuals.
    Returns the full spread DataFrame (all dates).
    """
    df = pd.read_csv(filename, index_col=0)
    df.index = date_index[:len(df)]
    return df

def load_spread_val(a, b):
    for fname in [f'{SPREADS_DIR}{a}_{b}.csv', f'{SPREADS_DIR}{b}_{a}.csv']:
        if os.path.exists(fname):
            df = pd.read_csv(fname, index_col=0, parse_dates=True)
            df.index = pd.to_datetime(df.index)
            df = df[(df.index >= VAL_START) & (df.index <= VAL_END)]
            if 'final_signal' not in df.columns:
                df['final_signal'] = df['signal']
            return df
    return None
```

```python
print(f'train: {len(train_returns)}')
print(f'val:   {len(val_returns)}')
print(f'combined: {len(date_index)}')
print(f'spread CSV: {len(pd.read_csv(f"{SPREADS_DIR}HPQ_WY.csv", index_col=0))}')
```

## Backtest functions

```python
def backtest_pair(df):
    """
    Simulate trading on pre-filtered df (already cut to the desired period).
    Uses final_signal column for entry/exit.
    """
    trades      = []
    position    = 0
    entry_price = None
    entry_date  = None

    for date, row in df.iterrows():
        signal = row['final_signal']
        spread = row['spread']

        if position == 0:
            if signal != 0:
                position    = signal
                entry_price = spread
                entry_date  = date
        else:
            if signal == 0 or signal != position:
                pnl = position * (spread - entry_price)
                trades.append({
                    'entry_date':   entry_date,
                    'exit_date':    date,
                    'direction':    position,
                    'entry_price':  entry_price,
                    'exit_price':   spread,
                    'pnl':          pnl,
                    'holding_days': (date - entry_date).days,
                })
                if signal != 0:
                    position    = signal
                    entry_price = spread
                    entry_date  = date
                else:
                    position    = 0
                    entry_price = None
                    entry_date  = None

    return pd.DataFrame(trades)


def backtest_baseline(df, entry_z=1.0, exit_z=0.3):
    """
    Naive baseline: fixed z-score thresholds, no regime filter, no MinHash.
    Overwrites final_signal with simple threshold rule.
    """
    df = df.copy()
    df['final_signal'] = 0
    df.loc[df['z_score'] < -entry_z, 'final_signal'] =  1
    df.loc[df['z_score'] >  entry_z, 'final_signal'] = -1
    df.loc[df['z_score'].abs() < exit_z, 'final_signal'] = 0
    return backtest_pair(df)


def compute_metrics(trades_df):
    if trades_df.empty:
        return {}

    pnl = trades_df['pnl']

    cumulative   = pnl.cumsum()
    rolling_max  = cumulative.cummax()
    max_drawdown = (cumulative - rolling_max).min()

    return {
        'total_pnl':        round(pnl.sum(), 4),
        'n_trades':         len(pnl),
        'win_rate':         round((pnl > 0).mean(), 4),
        'avg_pnl':          round(pnl.mean(), 4),
        'avg_win':          round(pnl[pnl > 0].mean(), 4) if (pnl > 0).any() else 0,
        'avg_loss':         round(pnl[pnl < 0].mean(), 4) if (pnl < 0).any() else 0,
        'sharpe':           round(pnl.mean() / pnl.std() * np.sqrt(252), 4) if pnl.std() > 0 else 0,
        'max_drawdown':     round(max_drawdown, 4),
        'avg_holding_days': round(trades_df['holding_days'].mean(), 1),
    }
```

```python
def compute_metrics(trades_df):
    if trades_df.empty:
        return {}
    pnl = trades_df['pnl']
    cumulative   = pnl.cumsum()
    rolling_max  = cumulative.cummax()
    max_drawdown = (cumulative - rolling_max).min()
    return {
        'total_pnl':        round(pnl.sum(), 4),
        'n_trades':         len(pnl),
        'win_rate':         round((pnl > 0).mean(), 4),
        'avg_pnl':          round(pnl.mean(), 4),
        'avg_win':          round(pnl[pnl > 0].mean(), 4) if (pnl > 0).any() else 0,
        'avg_loss':         round(pnl[pnl < 0].mean(), 4) if (pnl < 0).any() else 0,
        'sharpe':           round(pnl.mean() / pnl.std() * np.sqrt(252), 4) if pnl.std() > 0 else 0,
        'max_drawdown':     round(max_drawdown, 4),
        'avg_holding_days': round(trades_df['holding_days'].mean(), 1),
    }

def backtest_baseline(df, entry_z=1.0, exit_z=0.3):
    df = df.copy()
    df['final_signal'] = 0
    df.loc[df['z_score'] < -entry_z, 'final_signal'] =  1
    df.loc[df['z_score'] >  entry_z, 'final_signal'] = -1
    df.loc[df['z_score'].abs() < exit_z, 'final_signal'] = 0
    return backtest_pair(df)
```

## Run across all pairs

```python
all_trades     = []
baseline_trades = []
summary_rows   = []

for _, row in cointegration_pairs.iterrows():
    a, b = row['asset_a'], row['asset_b']
    df = load_spread_val(a, b)
    print(f'{a}/{b}: df={df is not None}, len={len(df) if df is not None else 0}')
    break  # just check first pair
    if df is None or df.empty:
        continue

    # strategy
    trades_df = backtest_pair(df)
    if not trades_df.empty:
        trades_df['asset_a'] = a
        trades_df['asset_b'] = b
        all_trades.append(trades_df)
        m = compute_metrics(trades_df)
        m['asset_a'] = a
        m['asset_b'] = b
        summary_rows.append(m)

    # baseline
    base_df = backtest_baseline(df)
    if not base_df.empty:
        base_df['asset_a'] = a
        base_df['asset_b'] = b
        baseline_trades.append(base_df)

all_trades_df = pd.concat(all_trades,      ignore_index=True) if all_trades      else pd.DataFrame()
baseline_df   = pd.concat(baseline_trades, ignore_index=True) if baseline_trades else pd.DataFrame()
summary_df    = pd.DataFrame(summary_rows).sort_values('total_pnl', ascending=False) if summary_rows else pd.DataFrame()

print(f'Strategy: {len(all_trades_df)} trades across {len(summary_df)} pairs')
print(f'Baseline: {len(baseline_df)} trades')
```

## Portfolio metrics

```python
portfolio_metrics = compute_metrics(all_trades_df)
baseline_metrics  = compute_metrics(baseline_df)

comparison = pd.DataFrame({
    'strategy': portfolio_metrics,
    'baseline': baseline_metrics,
})
print('Validation: Strategy vs Baseline')
print(comparison)
```

## Cumulative PnL — strategy vs baseline

```python
print(f'all_trades list: {len(all_trades)}, baseline list: {len(baseline_trades)}')
```

```python
strategy_cum = all_trades_df.groupby('exit_date')['pnl'].sum().cumsum()
baseline_cum = baseline_df.groupby('exit_date')['pnl'].sum().cumsum()

plt.figure(figsize=(12, 4))
plt.plot(strategy_cum.index, strategy_cum.values, label='Strategy (GMM + MinHash)')
plt.plot(baseline_cum.index, baseline_cum.values, label='Baseline (fixed threshold)', linestyle='--')
plt.axhline(0, color='gray', linewidth=0.5)
plt.title('Validation: Strategy vs Baseline — Cumulative PnL')
plt.xlabel('Date')
plt.ylabel('Cumulative PnL (spread units)')
plt.legend()
plt.tight_layout()
plt.show()
```
